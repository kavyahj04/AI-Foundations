### Concept 1: "is-a" → inheritance

##### use inheritance when the child genuinely is a more specific version of the parent 

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} says Woof!")

d = Dog("Rex")
d.eat()
d.bark()
print("isinstance(d, Animal):", isinstance(d, Animal))
```

### Concept 2: "has-a" → composition

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        print(f"  Engine starting with {self.horsepower} HP")

class Car:
    def __init__(self, make, engine):
        self.make = make
        self.engine = engine     # HAS-A: Car holds an Engine as an attribute

    def start(self):
        print(f"{self.make} is starting...")
        self.engine.start()      # delegates to the engine's own behavior

my_engine = Engine(300)
my_car = Car("Toyota", my_engine)

my_car.start()
print("isinstance(my_car, Engine):", isinstance(my_car, Engine))
print("my_car.engine is my_engine:", my_car.engine is my_engine)
```

# Class and Instance Basics

#### one class → as many independent instances as you want, each living at its own spot in memory, each holding its own data.

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance  = balance
    
    def deposit(self, amount):
        self.balance += amount
```

```python
acc1 = Account("Kavya", 100)

acc2 = Account("Mayur", 75)
```

```python
print("acc1 itself:", acc1)
print("acc1.owner:", acc1.owner)
print("acc1.balance:", acc1.balance)
print("id(acc1):", id(acc1))
```

```python
print("acc2 itself:", acc2)
print("acc2.owner:", acc2.owner)
print("acc2.balance:", acc2.balance)
print("id(acc2):", id(acc2))
```

#### What actually happens when you write Account("Kavya", 100)

##### It looks like one action, but it's really two steps. I made those two steps print out loud so you can see them happen:

```python
# class Account:
#     def __new__(cls, *args, **kwargs):
#         print("Step 1 (__new__): allocate raw, empty memory for a new object")
#         instance = super().__new__(cls)
#         print(f"  -> empty object created at id={id(instance)}")
#         return instance

#     def __init__(self, owner, balance):
#         print("Step 2 (__init__): fill in the data on that object")
#         self.owner = owner
#         self.balance = balance

# acc1 = Account("Kavya", 100)
```

#### __new__ runs first. Python reserves a blank spot in memory for the object — it exists now, has an id, but no owner or balance yet.

#### __init__ runs second, on that same blank object. This is the part you actually write normally (def __init__(self, owner, balance):) — it takes the empty object and fills in its starting data.



#### type() and isinstance()

```python
acc1 = Account("Kavya", 100)

print("type(acc1):", type(acc1))
print("type(acc1) == Account:", type(acc1) == Account)

print("isinstance(acc1, Account):", isinstance(acc1, Account))
print("isinstance(acc1, str):", isinstance(acc1, str))
```

### type(acc1) tells you exactly which class built this object — Account.
### isinstance(acc1, Account) asks the same thing, but as a yes/no question: "is this an Account?" → True

## Self

#### self is nothing but an instance
#### instance will have only 2 info — owner, balance - in this case

### self = "the specific account this method was called on." Inside the method, self.balance means "that account's balance" — not some global balance, not any other account's balance.

#### if you're writing def something(...) inside a class, self goes in your parameter list. If you're calling something_instance.method(...), you never type self — you just give the other arguments.

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount, note):
        print("self  ->", self.owner)
        print("amount ->", amount)
        print("note   ->", note)
        self.balance += amount

acc1 = Account("Kavya", 100)
acc1.deposit(50, "salary credit")
```

#### When you call acc1.deposit(50, "salary credit")
##### Seat 1 (self) is filled automatically by Python with acc1, because you called it with acc1.
##### Seat 2 (amount) is filled by the first thing you typed in the parentheses: 50
##### Seat 3 (note) is filled by the second thing you typed: "salary credit"


### Instances never interfere with each other

```python
acc1 = Account("Kavya", 100)
acc2 = Account("Rahul", 100)

print("Before:")
print("  acc1:", acc1.owner, acc1.balance)
print("  acc2:", acc2.owner, acc2.balance)

acc1.deposit(500)

print("After acc1.deposit(500):")
print("  acc1:", acc1.owner, acc1.balance)
print("  acc2:", acc2.owner, acc2.balance)
```
